```python
import yfinance as yf
stock_TSLA = yf.Ticker("TSLA")
info = stock_TSLA.info

print(info['longName'])
print(info['sector'])
print(info['industry'])
print(info['currency'])
print(info['website'])
print(info['fullTimeEmployees'])
print(info['currentPrice'])
print(info['longBusinessSummary'][:100])
```

```python
history = stock_TSLA.history(
    interval="1d",
    period="5d",
    auto_adjust=False
)
print(history)
```
